```python
import sys
import os
sys.path.append(os.path.expanduser('~/Desktop'))
import relations
import numpy as np
```

This script is to write the txt files that contain the zero points and dust correction numbers for all JWST+HST filters. Steve has given me the zero points and E(B-V) for the field, and we compute the dust correction according to the CCM89 Milky Way attenuation curve here. We output a txt file with zp and dustcorr columns for use in another script to correct pixel flux values.

```python
# import filters with zp
GS_zp_tab = np.genfromtxt('GOODSS_v095_zp.txt', dtype=None, encoding='utf-8', names=['filter', 'zp'])
GN_zp_tab = np.genfromtxt('GOODSN_v095_zp.txt', dtype=None, encoding='utf-8', names=['filter', 'zp'])
```

```python
GS_filters_names = GS_zp_tab['filter']
GN_filters_names = GN_zp_tab['filter']
```

```python
GS_filters_microns = [0.435, 0.606, 0.775, 0.814, 0.850, 0.98, 1.05, 1.25, 1.40, 1.60, 0.70, 0.90, 1.15, 1.50, 1.62, 1.82, 2.00, 2.10, 2.50, 2.77, 3.00, 3.35, 3.56, 4.10, 4.30, 4.44, 4.60, 4.80]
GN_filters_microns = [0.435, 0.606, 0.775, 0.814, 0.850, 0.70, 0.90, 1.15, 1.50, 1.62, 1.82, 2.00, 2.10, 2.77, 3.00, 3.35, 3.56, 4.10, 4.30, 4.44, 4.60]
```

```python
GS_EBV = 0.015
GN_EBV = 0.01
```

```python
GS_k = relations.CCM89_MWextinction(GS_filters_microns)
GN_k = relations.CCM89_MWextinction(GN_filters_microns)
```

```python
GS_dustcorr = 10**(0.4*GS_k*GS_EBV)
GN_dustcorr = 10**(0.4*GN_k*GN_EBV)
```

```python
with open('GOODSS_v095_zp.txt') as f:
    header_line = f.readline().strip()

with open('GOODSS_v095_zp-dustcorr.txt', 'w') as f:
    f.write(header_line + "\n")
    f.write("#filter  zp  GS_dustcorr\n")
    for filt, z, dc in zip(GS_filters_names, GS_zp_tab['zp'], GS_dustcorr):
        f.write(f"{filt}  {z:.7e}  {dc:.7f}\n")
```

```python
with open('GOODSN_v095_zp.txt') as f:
    header_line = f.readline().strip()

with open('GOODSN_v095_zp-dustcorr.txt', 'w') as f:
    f.write(header_line + "\n")
    f.write("#filter  zp  GN_dustcorr\n")
    for filt, z, dc in zip(GN_filters_names, GN_zp_tab['zp'], GN_dustcorr):
        f.write(f"{filt}  {z:.7e}  {dc:.7f}\n")
```
